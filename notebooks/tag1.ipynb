```python
def stromkostenJahr(verbrauch, preis):
    kostenProJahr = verbrauch * preis
    return round(kostenProJahr, 2)

def stromkostenMonat(verbrauch, preis):
    kostenProMonat = verbrauch / 12 * preis
    return round(kostenProMonat, 2)

verbrauch = float(input("Jährlicher Verbrauch in kWh: ").replace(",", "."))
preis = float(input("Preis pro kWh in €: ").replace(",", "."))

jahr = stromkostenJahr(verbrauch, preis)
monat = stromkostenMonat(verbrauch, preis)

print(f"Kosten pro Jahr: {jahr} €")
print(f"Kosten pro Monat: {monat} €")

if monat >= 80:
    print("Das ist zu teuer! Anbieter wechseln!")
else:
    print("Dein Strom ist günstig")
```
